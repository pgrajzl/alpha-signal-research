### Basic Starter Signals

##### *Simple Momentum*

Effectively the Law of Inertia: stocks that have been rising tend to keep rising (over the medium term), and stocks that have been falling tend to keep falling (over the medium term). This is one of the most well-documented anomalies in academic finance, persistent enough that it's a standard factor in most multi-factor models.

---
##### *Simple Mean Reversion*

Pretty much the opposite of momentum; stocks that experienced large drops will recover, whereas stocks that have experienced euphoric rallies will correct.

---
##### *Volume Spikes*

High volume could signal a variety of factors - perhaps more informed buying/selling, or perhaps more attention leading to short-term mispricings.

---

Note that all of these preliminary signal ideas are very naive and certainly not useful on their own. We'll need to get more creative to derive stronger insights.

### Basic Macro Starter Signals

##### *Rate Sensitivity Signal (Naive)*

The idea is that stocks respond differently to interest rate moves depending on their rate sensitivity — long-duration growth names and rate-sensitive sectors (utilities, REITs, high-multiple tech) tend to move more sharply with yield changes than defensive, low-duration names. This signal tries to capture that directly.

*Definition:* for each stock, compute its historical sensitivity to 10-year Treasury yield changes (a "rate beta," estimated via rolling regression of the stock's returns against changes in the 10Y yield). Multiply that rate beta by the recent change in the 10Y yield itself. The loose logic here is that a stock's expected reaction to recent yield moves is roughly equivalent to the rate beta * yield change.

*Weakness to note here:*
- Rate beta will need a rolling estimation window, and the choice of window length will likely have large impacts on IC (we'll test multiple combinations) - same for forward horizon
- No attempt yet to align FRED's yield data release timing precisely with stock return timing
- Very possible this ends up capturing something contemporaneous/mechanical rather than predictive, in which case it'll show up as a poor or misleading IC and needs rethinking
---

Note that all of these preliminary signal ideas are very naive and certainly not useful on their own. We'll need to get more creative to derive stronger insights.
