### Basic Starter Signals

##### *Simple Momentum*

**The idea:** stocks that have been rising steadily tend to keep rising, at least over the medium term. This is one of the most well-documented anomalies in academic finance — persistent enough that it's a standard factor in most multi-factor models.

---
##### *Simple Mean Reversion*

**Definition:** the stock's return over the trailing 12 months, excluding the most recent month.

---
##### *Volume Spikes*

**Why exclude the most recent month:** the most recent month's return tends to partially reverse (short-term reversal effect), so including it would blend two opposing signals together. Stripping it out isolates the pure "medium-term trend" effect from the "recent bounce/pullback" noise.

---

Note that all of these preliminary signal ideas are very naive and certainly not useful on their own. We'll need to get more creative to derive stronger insights.
