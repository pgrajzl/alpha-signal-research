### Basic Starter Signals

##### *Simple Momentum*

Effectively the Law of Inertia: stocks that have been rising tend to keep rising (over the medium term), and stocks that have been falling tend to keep falling (over the medium term). This is one of the most well-documented anomalies in academic finance, persistent enough that it's a standard factor in most multi-factor models.

---
##### *Simple Mean Reversion*

Pretty much the opposite of momentum; stocks that experienced large drops will recover, whereas stocks that have experienced euphoric rallies will correct.

---
##### *Volume Spikes*

High volume could signal a variety of factors - perhaps more informed buying/selling, or perhaps more attention leading to short-term mispricings.

---

Note that all of these preliminary signal ideas are very naive and certainly not useful on their own. We'll need to get more creative to derive stronger insights.
